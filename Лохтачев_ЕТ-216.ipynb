```python
import numpy as np

# 1. Арифметические операции
mas1 = np.array([34, 45, 56, 0, -1])
mas2 = np.array([10, 20, 30, 40, 50])
print("Сложение:", mas1 + mas2)
print("Вычитание:", mas1 - mas2)
print("Умножение:", mas1 * mas2)

# 2. Разные типы данных
print("\nВещественные:", np.array([1.5, 2.5]) + np.array([0.5, 1.0]))
print("Комплексные:", np.array([1+2j, 3+4j]) + np.array([1+2j, 3+4j]))
# Строки лучше делать обычными списками Python, без np.array
s_mas1 = ['Привет', 'мир']
s_mas2 = [', ', '!']
print("Строковые (сложение):", s_mas1[0] + s_mas2[0] + s_mas1[1] + s_mas2[1])

# 3. Тригонометрические функции
x1 = np.array([0.0, 1.0, 2.0])
x2 = np.array([0.5, 1.5, 2.5])
print("\nТригонометрия (sin+cos):", np.sin(x1) + np.cos(x2))

# 4. Массивы нулей и единиц
print("Массив из 6 нулей:", np.zeros(6, dtype=int))
print("Массив из 20 единиц:", np.ones(20, dtype=int))

# 5. Индексация
A = np.array([5, -3, 12, -8, 0, 7, -1, 4])
print("\nМассив A в обратном порядке:", A[::-1])
print("Сумма положительных элементов A:", np.sum(A[A > 0]))

# Месяцы
months = ["январь", "февраль", "март", "апрель", "май", "июнь",
          "июль", "август", "сентябрь", "октябрь", "ноябрь", "декабрь"]
print("Весенние месяцы:", months[2], months[3], months[4])
```

```python
import numpy as np

n = 30
np.random.seed(42) # Фиксируем случайные числа

# 1. Данные за 30 дней
vylet = np.random.randint(5, 20, n)
prilet = np.random.randint(5, 20, n)
print("Вылеты:", vylet)
print("Прилеты:", prilet)

# 2. Общее количество рейсов
total = vylet + prilet
print("Всего рейсов за каждый день:", total)

# 3. Минимумы и максимумы
print("\nМакс. вылетов:", np.max(vylet), " Мин. вылетов:", np.min(vylet))
print("Макс. прилетов:", np.max(prilet), " Мин. прилетов:", np.min(prilet))
print("Макс. всего:", np.max(total), " Мин. всего:", np.min(total))

# 4. Статистика
print("\nСтатистика по всем рейсам (total):")
print("Среднее:", np.mean(total))
print("Медиана:", np.median(total))
print("Дисперсия:", np.var(total))
print("Стандартное отклонение:", np.std(total))

# 5. В какую половину месяца было больше рейсов
half1 = np.sum(total[0:15])
half2 = np.sum(total[15:30])
if half1 > half2:
    print("\nВ первой половине месяца рейсов было больше (", half1, ">", half2, ")")
else:
    print("\nВо второй половине месяца рейсов было больше (", half2, ">", half1, ")")

# 6. Прилеты для суббот
# (1-й день понедельник имеет индекс 0, значит суббота имеет индекс 5. Шаг 7 дней)
print("\nПрилеты в субботы (дни 6, 13, 20, 27):", prilet[5::7])

# 7. Фильтрация вылетов
print("Дни, когда вылетов больше 10:", vylet[vylet > 10])
print("Дни, когда вылетов больше 10, но меньше 50:", vylet[(vylet > 10) & (vylet < 50)])
```

```python
import numpy as np

# 1. Создаем матрицу A 4x4 со случайными числами от 1 до 10
A = np.random.randint(1, 11, (4, 4))
print("Матрица A:\n", A)

# 2. Сумма четных элементов
print("\nСумма четных элементов A:", np.sum(A[A % 2 == 0]))

# 3. Вычисляем матрицы (T транспонирование, -1 обратная матрица)
A_T = A.T
A_inv = np.linalg.inv(A)
print("\nТранспонированная матрица A^T:\n", A_T)
print("\nОбратная матрица A^(-1):\n", np.round(A_inv, 4))

# 4. Находим определитель матрицы A (округляем для красивого вывода)
det_A = np.linalg.det(A)
print("\nОпределитель матрицы A:", round(det_A, 4))

# 5. Для матрицы A строим матрицу миноров M
M = np.zeros((4, 4)) # Создаем пустую матрицу 4x4
for i in range(4): # Идем по строкам
    for j in range(4): # Идем по столбцам
        # Вычеркиваем i-ю строку (axis=0) и j-й столбец (axis=1) из матрицы А
        temp = np.delete(A, i, axis=0)
        temp = np.delete(temp, j, axis=1)
        # Считаем определитель оставшейся части и округляем
        M[i, j] = round(np.linalg.det(temp), 4)

print("\nМатрица миноров M (для матрицы A):\n", M)
```

```python
import numpy as np

n = 30
np.random.seed(42) # Те же самые числа, что и в предыдущей ячейке

# 1. Данные за 30 дней
vylet = np.random.randint(5, 20, n)
prilet = np.random.randint(5, 20, n)

# Объединяем в двумерный массив (30 строк, 2 столбца)
airport = np.column_stack((vylet, prilet))
print("Массив (Вылет, Прилет):\n", airport)

# 2. Добавляем третий столбец (общее количество рейсов)
total = airport[:, 0] + airport[:, 1]
# Добавляем total как новый (третий) столбец к матрице airport
airport = np.column_stack((airport, total))
print("\nМассив с третьим столбцом (Вылет, Прилет, Всего):\n", airport)

# 3. Минимумы и максимумы для всех трёх показателей (по столбцам, axis=0)
print("\nМаксимумы по столбцам (Вылет, Прилет, Всего):", np.max(airport, axis=0))
print("Минимумы по столбцам (Вылет, Прилет, Всего):", np.min(airport, axis=0))

# 4. Оценить средние, медианы, дисперсии, отклонения (для 3-го столбца)
print("\nСтатистика по 3-му столбцу (Всего):")
print("Среднее:", np.mean(airport[:, 2]))
print("Медиана:", np.median(airport[:, 2]))
print("Дисперсия:", np.var(airport[:, 2]))
print("Стандартное отклонение:", np.std(airport[:, 2]))

# 5. В какую половину месяца было больше суммарное количество рейсов
half1 = np.sum(airport[0:15, 2])
half2 = np.sum(airport[15:30, 2])

if half1 > half2:
    print("\nВ первой половине месяца рейсов было больше (", half1, ">", half2, ")")
else:
    print("\nВо второй половине месяца рейсов было больше (", half2, ">", half1, ")")

# 6. Распечатать количество прилетов для суббот (используем срез)
print("\nПрилеты в субботы:", airport[5::7, 1])

# 7. Распечатать данные о ВЫЛЕТАХ (границы можно скорректировать)
vylet_col = airport[:, 0]

# когда их было больше 10
print("\nДни, когда вылетов больше 10:", vylet_col[vylet_col > 10])

# когда их было больше 10, но меньше 50
print("Дни, когда вылетов больше 10, но меньше 50:", vylet_col[(vylet_col > 10) & (vylet_col < 50)])
```
